# Task 01: Strings as Text Storage

A **string (`str`)** is used to store text in Python. In NLP, strings are very important because raw sentences and documents are normally stored as strings before we clean or tokenize them.

## Part (a)

We store the given sentence in a variable called `sentence`.

Then:
- `print()` displays the sentence.
- `type()` tells us the data type.
- `len()` counts the total number of characters, including spaces.

```python

sentence = "Artificial Intelligence is transforming Natural Language Processing"


print("Sentence:", sentence)


print("Data type:", type(sentence))

# Count and print the total number of characters
print("Total characters:", len(sentence))
```

## Part (b)

Python uses **indexing and slicing** to access parts of a string.

- `sentence[:10]` gives the first 10 characters, which form the first word.
- `sentence[-1]` gets the last character.
- `sentence[37:45]` extracts `"Language"` from this particular sentence.

```python

first_word = sentence[:10]

last_character = sentence[-1]

language_word = sentence[37:45]

print("First word:", first_word)
print("Last character:", last_character)
print("Word 'Language':", language_word)
```

## Part (c)

The `in` operator checks whether a piece of text exists inside another string.

The `find()` method returns the starting index of the first occurrence. If the text is not found, it returns `-1`.

```python

exists = "Intelligence" in sentence


index_position = sentence.find("Intelligence")

print("Does 'Intelligence' exist?", exists)
print("Starting index:", index_position)
```

## Part (d)

A **multi-line string** can be created using triple quotes.

The `split("\n")` method separates the paragraph wherever a newline occurs. This gives us a list containing the individual sentences.

```python

paragraph = """Natural Language Processing works with human language.
Tokenization breaks text into smaller units.
Machine learning can be used for many NLP tasks."""

# Split the paragraph at every newline
sentences = paragraph.split("\n")

# Print each sentence on a separate line
print("Sentences:")
for s in sentences:
    print(s)
```

# Task 02: Lists for Token Management

A **list** stores multiple values in an ordered and changeable collection. In NLP, lists are commonly used to store tokens (words) after tokenization.

## Part (a)

We create the required `tokens` list and use:
- `len(tokens)` for the number of elements
- `tokens[2]` for index 2
- `tokens[-1]` for the last element

```python

tokens = ["deep", "learning", "helps", "machines", "understand", "text"]

print("Full list:", tokens)
print("Length of list:", len(tokens))
print("Element at index 2:", tokens[2])
print("Last element:", tokens[-1])
```

## Part (b)

List methods allow us to modify a list:
- `append()` adds an item at the end.
- `insert()` adds an item at a specific index.
- `remove()` removes an item by its value.

```python

tokens.append("efficiently")
print("After append:", tokens)


tokens.insert(3, "computers")
print("After insert:", tokens)


tokens.remove("helps")
print("After remove:", tokens)
```

## Part (c)

The `+` operator can concatenate (join) two lists.

Slicing with `[-3:]` selects the last three elements.

```python
tokens_A = ["NLP", "uses", "Python"]
tokens_B = ["for", "text", "analysis"]

# Concatenate the two lists
full_tokens = tokens_A + tokens_B

print("Combined list:", full_tokens)
print("Last three elements:", full_tokens[-3:])
```

## Part (d)

First, `sort()` can arrange the list alphabetically.

For sorting by word length, `sorted()` with `key=len` is useful. In NLP, length-based sorting can be useful when we want to inspect vocabulary by token size, while alphabetical sorting is useful for dictionary-style organization.

```python

sorted_tokens = sorted(full_tokens)


length_sorted = sorted(full_tokens, key=len)

print("Alphabetically sorted:", sorted_tokens)
print("Sorted by word length:", length_sorted)

```

# Task 03: Dictionaries for Word Mapping

A **dictionary** stores data as key-value pairs.

In NLP, dictionaries are useful for:
- word frequency tables
- vocabulary-to-ID mappings
- storing linguistic information

## Part (a)

We create `word_freq` with each word as a key and its frequency as the value.

A value can be retrieved using its key, and dictionary values can be updated.

```python

word_freq = {
    "machine": 4,
    "learning": 7,
    "data": 5,
    "model": 3,
    "text": 6
}

print("Original dictionary:", word_freq)

print("Frequency of 'learning':", word_freq["learning"])

# Update the frequency of "data"
word_freq["data"] = 9
print("After updating 'data':", word_freq)
```

## Part (b)

We can add new key-value pairs simply by assigning values to new keys.

The `del` keyword removes an entry.

The `in` operator checks whether a key exists in the dictionary.

```python

word_freq["neural"] = 8
word_freq["network"] = 5


del word_freq["model"]


if "text" in word_freq:
    print("The key 'text' exists in the dictionary.")
else:
    print("The key 'text' does not exist.")

print("Final dictionary:", word_freq)
```

## Part (c)

A vocabulary often maps words to numeric IDs.

Here we create `vocab`, then reverse the mapping so that IDs become keys and words become values.

```python

vocab = {
    "nlp": 0,
    "python": 1,
    "token": 2,
    "vector": 3,
    "corpus": 4
}


id_to_word = {id_value: word for word, id_value in vocab.items()}

print("Word to ID:", vocab)
print("ID to word:", id_to_word)
```

## Part (d)

The `.items()` method gives both the key and value.

We loop through the dictionary and print each word with its frequency. The number of unique words is the dictionary's length.

```python

for word, frequency in word_freq.items():
    print(f'Word: "{word}" → Frequency: {frequency}')

print("Total unique words:", len(word_freq))
```

# Task 04: Sets for Unique Vocabulary

A **set** stores unique values. It automatically removes duplicate elements.

Sets are useful in NLP when we need a unique vocabulary or want to perform operations such as union, intersection, and difference.

## Part (a)

The original list contains repeated words. Converting it to a set removes duplicates automatically because a set can contain only one copy of each value.

```python
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]

vocab = set(raw)

print("Raw list:", raw)
print("Unique vocabulary:", vocab)
print("Number of unique words:", len(vocab))

```

## Part (b)

Set operations:
- **Union (`|`)** gives all elements from both sets.
- **Intersection (`&`)** gives only common elements.
- **Difference (`-`)** gives elements present in the first set but not the second.

```python
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

# Union: all unique elements from A and B
union_set = set_A | set_B

# Intersection: common elements
intersection_set = set_A & set_B

# Difference: elements only in A
difference_set = set_A - set_B

print("Union:", union_set)
print("Intersection:", intersection_set)
print("A - B:", difference_set)
```

## Part (c)

Stopwords are common words that are often removed during text preprocessing.

Here we use set difference to remove all words that occur in the `stopwords` set.

```python
stopwords = {"is", "the", "a", "in", "of", "and", "to"}

tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

# Convert tokens to a set and remove stopwords
cleaned_tokens = set(tokens) - stopwords

print("Cleaned tokens:", cleaned_tokens)
```

## Part (d)

`add()` inserts one new item into a set.

`discard()` removes an item if it exists. Unlike `remove()`, `discard()` does not raise an error if the item is missing.

The `in` operator checks membership.

```python
vocab.add("transformer")
vocab.add("bert")

vocab.discard("python")

print("Is 'ai' present?", "ai" in vocab)
print("Final vocabulary:", vocab)
```

# Task 05: Tuples for Fixed Linguistic Data

A **tuple** is an ordered collection that cannot normally be changed after creation.

Tuples are useful for fixed pairs such as `(word, POS-tag)` because the word-tag relationship should remain together.

## Part (a)

Each item in `pos_tags` is a tuple containing a word and its part-of-speech tag.

The second tuple is at index `1`, so:
- `pos_tags[1][0]` gives the word
- `pos_tags[1][1]` gives the tag

```python
# List of word-POS tag tuples
pos_tags = [
    ("language", "NOUN"),
    ("processes", "VERB"),
    ("human", "ADJ"),
    ("text", "NOUN"),
    ("quickly", "ADV")
]

print("POS tag list:", pos_tags)

# Access the second tuple
print("Word from second tuple:", pos_tags[1][0])
print("Tag from second tuple:", pos_tags[1][1])
```

## Part (b)

Tuple unpacking assigns the two values to separate variables.

Tuples are immutable, so trying to change an element raises a `TypeError`. The `try-except` block catches the error so the program continues running.

```python
# Unpack the first tuple
word, tag = pos_tags[0]

print("Word:", word)
print("Tag:", tag)

# Try to modify the first tuple
try:
    pos_tags[0][1] = "VERB"
except TypeError as e:
    print("Exact Python error message:", e)
```

## Part (c)

The function checks every `(word, tag)` pair.

If the tag is `"NOUN"`, the word is added to the result list.

```python
def extract_nouns(pos_list):
    # Create an empty list for noun words
    nouns = []

    # Check every word-tag pair
    for word, tag in pos_list:
        if tag == "NOUN":
            nouns.append(word)

    return nouns

# Call the function
noun_words = extract_nouns(pos_tags)

print("Nouns:", noun_words)
```

## Part (d)

The pipeline is stored in a tuple because the sequence of steps is fixed.

`index()` finds the position of `"tokenize"`.

Trying to delete a tuple element raises a `TypeError`, demonstrating tuple immutability.

```python
pipeline_steps = (
    "load_data",
    "clean_text",
    "tokenize",
    "remove_stopwords",
    "vectorize"
)

print("Pipeline:", pipeline_steps)
print("Index of 'tokenize':", pipeline_steps.index("tokenize"))
print("Total number of steps:", len(pipeline_steps))

# Demonstrate that tuples cannot be modified
try:
    del pipeline_steps[1]
except TypeError as e:
    print("Error while deleting an element:", e)
```

# Task 06: String Cleaning Methods

String cleaning is one of the basic steps in NLP preprocessing.

Important methods here are:
- `strip()` – removes spaces from the beginning and end
- `lower()` – converts text to lowercase
- `replace()` – replaces selected characters or text
- `split()` – converts text into tokens
- `join()` – combines items into one string

## Part (a)

We apply each method separately and store the result so we can see the effect of every preprocessing step.

```python
text = " MACHINE Learning is AMAZING!!! "

# Step 1: remove leading and trailing spaces
step1 = text.strip()
print("After strip():", step1)

# Step 2: convert to lowercase
step2 = step1.lower()
print("After lower():", step2)

# Step 3: remove all exclamation marks
step3 = step2.replace("!", "")
print("After replace():", step3)
```

## Part (b)

After cleaning, `split()` separates the string into words.

A list comprehension applies `upper()` to every word.

Finally, `join()` combines the uppercase words using `" | "` as the separator.

```python
# Tokenize the cleaned text
tokens = step3.split()
print("Tokens:", tokens)

# Convert every token to uppercase
uppercase_tokens = [word.upper() for word in tokens]
print("Uppercase tokens:", uppercase_tokens)

# Join the words using " | "
joined_text = " | ".join(uppercase_tokens)
print("Joined text:", joined_text)
```

## Part (c)

`replace()` changes every occurrence of `"Java"` into `"Python"`.

After replacement, `find()` returns the starting index of the first `"Python"`.

```python
sentence = "I love Java and Java is popular"

# Replace every occurrence of Java
updated_sentence = sentence.replace("Java", "Python")

# Find the first occurrence of Python
python_position = updated_sentence.find("Python")

print("Updated sentence:", updated_sentence)
print("First 'Python' index:", python_position)
```

## Part (d)

The `normalize()` function performs the required cleaning steps in order:
1. remove surrounding whitespace
2. convert to lowercase
3. remove commas
4. remove full stops

```python
def normalize(text):
    # Remove leading and trailing spaces
    text = text.strip()

    # Convert all letters to lowercase
    text = text.lower()

    # Remove commas and full stops
    text = text.replace(",", "")
    text = text.replace(".", "")

    return text

# Test the function on two messy strings
test1 = "  Hello, NLP World.  "
test2 = " Python, is Great for NLP. "

print("Cleaned test 1:", normalize(test1))
print("Cleaned test 2:", normalize(test2))
```

# Task 07: String Validation Methods

Validation methods help us check the type or pattern of text.

Here we use:
- `startswith()`
- `endswith()`
- `find()`
- `isalpha()`
- `isdigit()`

## Part (a)

`startswith()` checks the beginning of a string.

`endswith()` checks the ending of a string.

`find()` returns the index where `"future"` starts.

```python
text = "NLP is the future of AI research"


starts_nlp = text.startswith("NLP")

ends_research = text.endswith("research")

future_position = text.find("future")

print("Starts with 'NLP':", starts_nlp)
print("Ends with 'research':", ends_research)
print("Position of 'future':", future_position)
```

## Part (b)

`isalpha()` returns `True` only when all characters are alphabetic.

`isdigit()` returns `True` only when all characters are digits.

If neither condition is true, we label the token as **mixed**.

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        label = "purely alphabetic"
    elif token.isdigit():
        label = "purely numeric"
    else:
        label = "mixed"

    print(f"{token} -> {label}")
```

## Part (c)

We separate the words into three lists:
- `pure_words`
- `pure_numbers`
- `mixed_tokens`

The conditions are checked using `isalpha()` and `isdigit()`.

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)
```

## Part (d)

The function `validate_token()` returns one of three labels.

This makes the validation logic reusable for many tokens instead of writing the same `if-elif-else` code repeatedly.

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in test_tokens:
    print(token, "->", validate_token(token))
```

# Task 08: Functions and Lambda for NLP Pipelines

Functions make code reusable.

A **lambda function** is a small anonymous function, useful when the operation is short.

In NLP, separate functions can be combined into a preprocessing pipeline.

## Part (a)

The normal function and lambda function perform the same two operations:
1. remove surrounding whitespace
2. convert the text to lowercase

```python
def clean(text):
    # Remove surrounding spaces and convert to lowercase
    return text.strip().lower()

# Equivalent one-line lambda function
clean_lambda = lambda text: text.strip().lower()

sample_text = " HELLO Natural Language Processing "

print("Regular function:", clean(sample_text))
print("Lambda function:", clean_lambda(sample_text))
```

## Part (b)

First, `tokenize()` splits the sentence into words.

Then `remove_short()` keeps only words whose length is **greater than** `n`.

The two functions are chained together.

```python
def tokenize(text):
    # Convert a sentence into a list of word tokens
    return text.split()

def remove_short(tokens, n):
    # Keep only tokens longer than n characters
    return [token for token in tokens if len(token) > n]

sentence = "I love working with NLP tools in Python daily"

# Chain the functions
tokens = tokenize(sentence)
long_tokens = remove_short(tokens, 3)

print("Tokens:", tokens)
print("Tokens longer than 3 characters:", long_tokens)
```

## Part (c)

The function creates a dictionary where:
- key = word
- value = length of the word

A dictionary comprehension is a simple way to build this mapping.

```python
def word_lengths(tokens):
    # Create a word -> length dictionary
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]

length_dictionary = word_lengths(tokens)

print("Word lengths:", length_dictionary)
```

## Part (d)

The complete pipeline performs four stages:

**clean → tokenize → remove short words → calculate word lengths**

The function returns the final dictionary.

```python
def nlp_pipeline(text, min_len):
    # Step 1: clean the text
    cleaned = clean(text)

    # Step 2: tokenize the cleaned text
    tokens = tokenize(cleaned)

    # Step 3: remove words that are not longer than min_len
    filtered_tokens = remove_short(tokens, min_len)

    # Step 4: create a dictionary of word lengths
    return word_lengths(filtered_tokens)

# Test the pipeline on two sentences
sentence1 = " Natural Language Processing is interesting "
sentence2 = " Python makes text processing easier "

print("Pipeline output 1:", nlp_pipeline(sentence1, 3))
print("Pipeline output 2:", nlp_pipeline(sentence2, 4))
```

# Task 09: map(), filter(), sorted() and Counter

These Python tools are very useful for NLP data processing.

- `map()` applies an operation to every item.
- `filter()` keeps items that satisfy a condition.
- `sorted()` orders items.
- `Counter` counts frequencies.

## Part (a)

We use `map()` with lambda functions.

The first operation converts every word to lowercase.

The second operation calculates the number of characters in each word.

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# Convert every word to lowercase
lowercase_words = list(map(lambda word: word.lower(), words))

# Calculate the length of every word
word_lengths_list = list(map(lambda word: len(word), words))

print("Lowercase words:", lowercase_words)
print("Word lengths:", word_lengths_list)
```

## Part (b)

`filter()` keeps only the elements for which the lambda condition returns `True`.

- `words_long` contains words longer than 3 characters.
- `words_with_t` contains words having the letter `t`, ignoring case.

```python
# Keep words with more than 3 characters
words_long = list(filter(lambda word: len(word) > 3, words))

# Keep words containing the letter "t"
words_with_t = list(filter(lambda word: "t" in word.lower(), words))

print("Words longer than 3 characters:", words_long)
print("Words containing 't':", words_with_t)
```

## Part (c)

`sorted()` creates a sorted version of the list.

We demonstrate:
1. default alphabetical order
2. ascending word length
3. descending word length

```python
# 1. Alphabetical order
alphabetical = sorted(words)

# 2. Word length, ascending
length_ascending = sorted(words, key=len)

# 3. Word length, descending
length_descending = sorted(words, key=len, reverse=True)

print("Alphabetical:", alphabetical)
print("Length ascending:", length_ascending)
print("Length descending:", length_descending)
```

## Part (d)

`Counter` from the `collections` module counts how many times each word occurs.

For the final statistics:
- `sum(freq.values())` gives the total number of words.
- `len(freq)` gives the number of unique words.

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"

# Split the sentence into words
tokens = sentence.split()

# Count the frequency of each word
freq = Counter(tokens)

print("Full Counter:", freq)

# Print the three most common words
print("3 most common words:", freq.most_common(3))

# Frequency of python
print("Frequency of 'python':", freq["python"])

# Total number of words
print("Total number of words:", sum(freq.values()))

# Number of unique words
print("Unique vocabulary size:", len(freq))
```
