

## Clean a Paragraph Using Core Python

**Task:** Write a function that returns a cleaned list of lowercase words with leading/trailing punctuation removed.

### Approach
We use only core Python. For each word:
1. split the paragraph into whitespace-separated pieces;
2. convert it to lowercase;
3. remove punctuation from the beginning and end with `strip()`.

```python
def clean_words(text):
    """Return lowercase words with surrounding punctuation removed."""

    # Punctuation that commonly appears around words.
    punctuation = ".,!?;:'\"()[]{}"

    cleaned = []

    for word in text.lower().split():
        # Remove punctuation only from the beginning/end.
        word = word.strip(punctuation)

        if word:  # Avoid adding empty strings.
            cleaned.append(word)

    return cleaned


new_paragraph = "NLP is useful! Python, however, needs practice. Text-processing is important."
print("Cleaned words:", clean_words(new_paragraph))
```

## Verify Activity 2 – Top 10 Words with `Counter`

**Task:** Find the 10 most frequent words in an article and display them as a clean word/count table.

### Explanation
This follows the same normalization idea as Activity 4. Lowercasing and removing punctuation prevents surface variations such as `Data` and `data.` from being counted separately.

```python
from collections import Counter
import re

article = """
Natural language processing allows computers to work with human language.
Language data can be noisy, so text processing is important. 
Text processing includes cleaning, tokenization, and normalization.
NLP models learn useful patterns from language data and text data.
"""

# Normalize the article before counting.
clean_article = re.sub(r"[^\w\s]", "", article.lower())
article_tokens = clean_article.split()

# Count all words and select the top 10.
top_10 = Counter(article_tokens).most_common(10)

print(f"{'Word':<15}{'Count':>5}")
print("-" * 20)
for word, count in top_10:
    print(f"{word:<15}{count:>5}")
```

## Verify Activity 3 – Sentence DataFrame with Word and Character Counts

**Task:** Build a DataFrame with one row per sentence and add `word_count` and `char_count` using `.apply()`.

### Explanation
`word_count` is calculated after splitting on whitespace. `char_count` counts the characters in the original sentence, including spaces and punctuation.

```python
import pandas as pd

paragraph = [
    "Natural language processing is an exciting field.",
    "Python makes text manipulation easier.",
    "Data is the foundation of NLP systems."
]

df_verify = pd.DataFrame({"sentence": paragraph})

# Count words and characters for each sentence.
df_verify["word_count"] = df_verify["sentence"].apply(lambda x: len(x.split()))
df_verify["char_count"] = df_verify["sentence"].apply(len)

print(df_verify.to_string(index=False))
```

## Verify Activity 4 – Cosine Similarity for Two Vector Pairs

**Task:** Compute cosine similarity for two pairs of word-count vectors and determine which pair has the larger similarity value.

### Important
The comparison is based on the numerical cosine-similarity values. A value closer to 1 means the vectors point in more similar directions. We are not comparing their raw lengths.

```python
import numpy as np

def cosine_similarity(a, b):
    """Calculate cosine similarity using NumPy."""
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))


# Pair 1: very similar word-count distribution.
pair1_a = np.array([3, 2, 1])
pair1_b = np.array([6, 4, 2])

# Pair 2: different word-count distribution.
pair2_a = np.array([3, 1, 2])
pair2_b = np.array([1, 5, 1])

similarity_1 = cosine_similarity(pair1_a, pair1_b)
similarity_2 = cosine_similarity(pair2_a, pair2_b)

print("Pair 1 cosine similarity:", similarity_1)
print("Pair 2 cosine similarity:", similarity_2)

if similarity_1 > similarity_2:
    print("Pair 1 has the larger cosine similarity.")
else:
    print("Pair 2 has the larger cosine similarity.")

print("Reason: the pair with the larger cosine value has more similar vector direction.")
```

## Verify Activity 5 – Reviews, `Counter`, DataFrame and Sorting

**Task:** Load 10 short reviews from a JSON structure, build a DataFrame, add:
- `word_count`
- a Counter-based `top_word` column for each review (ignoring case)

Finally, sort by `word_count` in descending order.

### Explanation
For every review, we:
1. normalize the text to lowercase;
2. split it into words;
3. count the words with `Counter`;
4. take the most frequent word;
5. store the result as a new DataFrame column.

```python
import json
from collections import Counter
import re
import pandas as pd


reviews = [
    {"review": "The product is excellent and the quality is excellent."},
    {"review": "Good service and very helpful staff."},
    {"review": "The delivery was fast and the packaging was good."},
    {"review": "Excellent product with excellent performance."},
    {"review": "The application is simple, clean, and easy to use."},
    {"review": "Fast delivery and good customer support."},
    {"review": "The quality is good and the price is reasonable."},
    {"review": "I am happy with the product and the service."},
    {"review": "Very useful product with a simple design."},
    {"review": "The service was excellent and the staff were helpful."}
]

# Save the sample reviews to a JSON file first.
with open("reviews.json", "w", encoding="utf-8") as f:
    json.dump(reviews, f, indent=4)

# Load the JSON file.
with open("reviews.json", "r", encoding="utf-8") as f:
    loaded_reviews = json.load(f)

# Create a DataFrame.
reviews_df = pd.DataFrame(loaded_reviews)

def get_top_word(text):
    """Return the most frequent word in one review."""
    cleaned = re.sub(r"[^\w\s]", "", text.lower())
    words = cleaned.split()

    if not words:
        return ""

    return Counter(words).most_common(1)[0][0]

# Add word count and most frequent word for every review.
reviews_df["word_count"] = reviews_df["review"].apply(
    lambda x: len(x.split())
)
reviews_df["top_word"] = reviews_df["review"].apply(get_top_word)

# Rank reviews from longest to shortest.
reviews_ranked = reviews_df.sort_values(
    by="word_count",
    ascending=False
)

print(reviews_ranked.to_string(index=False))
```
